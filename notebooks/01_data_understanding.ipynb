```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Path to the raw data folder
DATA_PATH = Path("../data/raw")

# Get all CSV files
csv_files = sorted(DATA_PATH.glob("*.csv"))

# Check the number of CSV files
print(f"Number of CSV files: {len(csv_files)}")

# Load all CSV files into a dictionary
data = {}

for file in csv_files:
    df_name = file.stem
    data[df_name] = pd.read_csv(file)

    print(f"Loaded: {file.name}")
```

```python
dataset_summary = pd.DataFrame([
    {
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    }
    for name, df in data.items()
])

dataset_summary
```

```python
for name, df in data.items():
    print(f"\n{'=' * 60}")
    print(f"Dataset: {name}")
    print(f"{'=' * 60}")

    display(df.head())
```

```python
for name, df in data.items():
    print(f"\n{'=' * 60}")
    print(f"Dataset: {name}")
    print(f"{'=' * 60}")

    df.info()
```

```python
for name, df in data.items():
    missing = pd.DataFrame({
        "Missing Values": df.isnull().sum(),
        "Missing Percentage": (
            df.isnull().sum() / len(df) * 100
        ).round(2)
    })

    missing = missing[missing["Missing Values"] > 0]

    print(f"\n{'=' * 60}")
    print(f"Missing Values: {name}")
    print(f"{'=' * 60}")

    if missing.empty:
        print("No missing values found.")
    else:
        display(missing)
```

```python
for name, df in data.items():
    duplicate_count = df.duplicated().sum()

    print(f"{name}: {duplicate_count} duplicate rows")
```

```python
for name, df in data.items():
    print(f"\n{name}")
    print(list(df.columns))
```

```python
# Check movement types and their frequency
data["inventory_movements"]["movement_type"].value_counts(
    dropna=False
)

# Investigate missing IDs by movement type
movement_missing_analysis = (
    data["inventory_movements"]
    .groupby("movement_type", dropna=False)
    .agg(
        Total_Records=("movement_id", "count"),
        Missing_PO_Line_ID=("purchase_order_line_id",
                            lambda x: x.isna().sum()),
        Missing_Transfer_ID=("transfer_id",
                             lambda x: x.isna().sum())
    )
    .reset_index()
)

movement_missing_analysis
```

```python
# Check purchase order statuses
data["purchase_orders"]["status"].value_counts(
    dropna=False
)

# Compare received_at values with order status
received_date_analysis = (
    data["purchase_orders"]
    .groupby("status", dropna=False)
    .agg(
        Total_Orders=("purchase_order_id", "count"),
        Missing_Received_At=("received_at",
                             lambda x: x.isna().sum())
    )
    .reset_index()
)

received_date_analysis["Missing_Percentage"] = (
    received_date_analysis["Missing_Received_At"]
    / received_date_analysis["Total_Orders"]
    * 100
).round(2)

received_date_analysis
```

```python
# Compare total rows, non-null IDs, and movement types
movements = data["inventory_movements"]

print("Total rows:", len(movements))
print("Non-null movement IDs:", movements["movement_id"].count())
print("Missing movement IDs:", movements["movement_id"].isna().sum())
print("Missing movement types:", movements["movement_type"].isna().sum())

# Count rows per movement type, including rows with missing values
movements.groupby(
    "movement_type",
    dropna=False
).size().reset_index(name="Row_Count")
```

## Key Findings & Initial Insights

### 1. Dataset Overview
- The project contains seven CSV datasets with a combined total of 243,165 rows and 41 columns.
- The datasets cover inventory movements, opening inventory balances, products, purchase order lines, purchase orders, suppliers, and warehouses.
- `inventory_movements` is the largest dataset, containing 138,436 records.

### 2. Missing Values
- Five datasets contain no missing values: `inventory_opening_balances`, `products`, `purchase_order_lines`, `suppliers`, and `warehouses`.
- In `inventory_movements`, `purchase_order_line_id` is missing for adjustment, sale, and transfer movements. `transfer_id` is missing for adjustment, receipt, and sale movements.
- The missing inventory movement IDs appear to follow the movement type. These values should be retained unless further validation identifies a data-quality issue.
- In `purchase_orders`, 873 records (4.85%) have a missing `received_at` value. All 406 cancelled orders and 467 open orders have no receipt date, while partial and received orders have a receipt date.
- The missing receipt dates appear consistent with order status and should not be filled arbitrarily.

### 3. Initial Relationship Observations
- Products can be linked to suppliers using `supplier_id`.
- Purchase order lines can be linked to purchase orders and products using `purchase_order_id` and `product_id`.
- Purchase orders contain supplier and warehouse identifiers.
- Inventory movements and opening balances contain product and warehouse identifiers, making it possible to analyse inventory by product and warehouse.
- Inventory movements may also link to purchase order lines or transfers, depending on the movement type.
- These relationships require validation before analysis.

### 4. Potential Areas for Analysis
- Inventory movement patterns by movement type, product, and warehouse.
- Purchase order fulfilment and outstanding orders.
- Supplier lead times and reliability.
- Inventory levels compared with product reorder points.
- Warehouse-level inventory distribution and capacity.
